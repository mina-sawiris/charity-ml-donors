```python
import pandas as pd

data = pd.read_csv('census.csv')

# 1. Check the relationship between education_level (text) and education-num
print("=== Checking relationship between education_level and education-num ===")

# Group by education level to find the minimum, maximum, and number of unique values for education-num
edu_check = data.groupby('education_level')['education-num'].agg(['min', 'max', 'nunique'])
print(edu_check)
print("\n" + "="*50 + "\n")

# 2. Print all categorical columns and their unique values
print("=== Categorical Columns and Their Unique Values ===")

# Filter the dataset to get only the columns with 'object' (text/categorical) data type
categorical_columns = data.select_dtypes(include=['object']).columns

for col in categorical_columns:
    unique_values = data[col].unique()
    print(f"Column: {col}")
    print(f"Number of unique options: {len(unique_values)}")
    print(f"Options are: {unique_values}")
    print("-" * 40)
```
